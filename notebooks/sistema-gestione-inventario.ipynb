# Sistema di gestione degli elementi

## Obiettivo

Realizzare un codice in Python che consenta agli utenti di creare e gestire una collezione di elementi specifici integrando funzionalità utili come la registrazione, la visualizzazione, la ricerca e l’analisi degli elementi.

```python
#Collezione principale
inventario = []

#ID progressivo
id_progressivo = 1

#Valori ammessi
condizioni_ammesse = ["nuovo", "ottimo", "buono", "usato"]
```

```python
def registra_elemento(inventario, id_progressivo, tipo, colore, taglia, prezzo, condizione, condizioni_ammese):

    """
    Registra un nuovo capo nell'inventario.

    Args:
        inventario (list[dict]): lista di capi già registrati
        id_progressivo (int): ID da assegnare al nuovo capo
        tipo (str): tipo di capo (es. "giacca")
        colore (str): colore del capo (es. "nero")
        taglia (str): taglia del capo (es. "M")
        prezzo (float): prezzo del capo
        condizione (str): condizione del capo (es. "nuovo")
        condizioni_ammese (list[str]): lista delle condizioni consentite

    Returns:
        tuple: (inventario aggiornato, nuovo id_progressivo, capo_creato)
    """
  
    #Normalizzazione input testuale
    tipo = tipo.strip().lower()
    colore = colore.strip().lower()
    taglia = taglia.strip().upper()
    condizione = condizione.strip().lower()

    #Validazione condizione
    if condizione not in condizioni_ammese:
        raise ValueError(f"Condizione non valida. Usa una tra: {condizioni_ammese}")

    #Validazione prezzo
    if prezzo < 0:
        raise ValueError("Il prezzo non può essere negativo.")

    #Controllo duplicati: stesso tipo+colore+taglia+prezzo+condizione
    for capo in inventario:
        if (capo["tipo"] == tipo and
            capo["colore"] == colore and
            capo["taglia"] == taglia and
            round(capo["prezzo"], 2) == round(prezzo, 2) and
            capo["condizione"] == condizione):
            raise ValueError("Capo duplicato: esiste già")

    #Creazione capo
    capo = {
        "id": id_progressivo,
        "tipo": tipo,
        "colore": colore,
        "taglia": taglia,
        "prezzo": float(prezzo),
        "condizione": condizione
    }

    inventario.append(capo)
    id_progressivo += 1

    return inventario, id_progressivo, capo
```

```python
inventario, id_progressivo, capo1 = registra_elemento(
    inventario=inventario,
    id_progressivo=id_progressivo,
    tipo="Giacca",
    colore="Nero",
    taglia="M",
    prezzo=19.99,
    condizione="Nuovo",
    condizioni_ammese=condizioni_ammesse
)

capo1
```

```python
def visualizza_elementi(inventario):
    """
    Visualizza tutti i capi presenti nell'inventario in modo leggibile
    Args:
        inventario (list[dict]): lista di capi registrati
    Returns:
        None
    """
    #Caso limite: inventario vuoto
    if not inventario:
        print("Inventario vuoto.")
        return

    #Stampa intestazione
    print("Elenco capi in inventario:")

    #Ciclo su tutti i capi
    for capo in inventario:
        print(
            "ID:", capo["id"],
            "| Tipo:", capo["tipo"],
            "| Colore:", capo["colore"],
            "| Taglia:", capo["taglia"],
            "| Prezzo:", capo["prezzo"], "€",
            "| Condizione:", capo["condizione"]
        )
```

```python
visualizza_elementi(inventario)
```

```python
def ricerca_elementi(inventario, tipo=None, condizione=None, colore=None, prezzo=None):
    """
    Ricerca elementi nell'inventario applicando uno o più filtri opzionali

    Args:
        inventario (list[dict]): lista dei capi registrati
        tipo (str, optional): tipo di capo da cercare
        condizione (str, optional): condizione del capo
        colore (str, optional): colore del capo
        prezzo (float, optional): prezzo esatto del capo

    Returns:
        list[dict]: lista dei capi che soddisfano i criteri
    """

    risultati = []  # Lista che conterrà i capi trovati

    #Ciclo su tutti i capi dell'inventario
    for capo in inventario:

        #Filtro per tipo 
        if tipo is not None and capo["tipo"] != tipo.lower():
            continue

        #Filtro per condizione
        if condizione is not None and capo["condizione"] != condizione.lower():
            continue

        #Filtro per colore
        if colore is not None and capo["colore"] != colore.lower():
            continue

        #Filtro per prezzo esatto
        if prezzo is not None and round(capo["prezzo"], 2) != round(prezzo, 2):
            continue

        #Se il capo supera tutti i filtri, lo aggiungo ai risultati
        risultati.append(capo)

    return risultati
```

```python
# Ricerca solo per tipo
ris = ricerca_elementi(inventario, tipo="giacca")
visualizza_elementi(ris)

# Ricerca combinata tipo + condizione
ris = ricerca_elementi(inventario, tipo="giacca", condizione="nuovo")
visualizza_elementi(ris)

# Ricerca completa
ris = ricerca_elementi(inventario, tipo="giacca", condizione="nuovo", colore="nero", prezzo=19.99)
visualizza_elementi(ris)
```

```python
#Test
ris = ricerca_elementi(inventario, tipo="maglione")
visualizza_elementi(ris)
```

```python
def statistiche(inventario):
    """
    Calcola e stampa statistiche sugli elementi presenti nell'inventario

    Statistiche incluse:
    - numero totale di capi
    - distribuzione per condizione
    - tipo più frequente
    - capo più recente (id più alto)

    Args:
        inventario (list[dict]): lista dei capi registrati

    Returns:
        None
    """
    #Caso limite: inventario vuoto
    if not inventario:
        print("Inventario vuoto.")
        return

    #Totale capi
    totale = len(inventario)
    print("Totale capi:", totale)

    #Distribuzione per condizione
    conteggio_condizioni = {}
    for capo in inventario:
        cond = capo["condizione"]
        conteggio_condizioni[cond] = conteggio_condizioni.get(cond, 0) + 1

    print("\nDistribuzione per condizione:")
    for cond, numero in conteggio_condizioni.items():
        print("-", cond, ":", numero)

    #Tipo più frequente
    conteggio_tipi = {}
    for capo in inventario:
        t = capo["tipo"]
        conteggio_tipi[t] = conteggio_tipi.get(t, 0) + 1

    tipo_piu_frequente = max(conteggio_tipi, key=conteggio_tipi.get)
    print("\nTipo più frequente:", tipo_piu_frequente, "(", conteggio_tipi[tipo_piu_frequente], ")")

    #Capo più recente (ID più alto)
    capo_piu_recente = max(inventario, key=lambda c: c["id"])
    print("\nCapo più recente:")
    print(
        "ID:", capo_piu_recente["id"],
        "| Tipo:", capo_piu_recente["tipo"],
        "| Colore:", capo_piu_recente["colore"],
        "| Taglia:", capo_piu_recente["taglia"],
        "| Prezzo:", capo_piu_recente["prezzo"], "€",
        "| Condizione:", capo_piu_recente["condizione"]
    )
```

```python
statistiche(inventario)
```
